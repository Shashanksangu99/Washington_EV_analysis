```python
#importing pandas library
import pandas as pd

#reading the EV csv file as pandas DataFrame
ev_df= pd.read_csv('/content/Electric_Vehicle_Population_Data_20240831.csv',index_col=0)

ev_df.shape
ev_df
```

```python
#Checking the columns in the dataset
#ev_df.columns

#Formatting the columns into SQL query friendly by adding '_'

ev_df.columns=ev_df.columns.str.lower()

ev_df.columns=ev_df.columns.str.replace(" ","_")


#Correcting VIN column
ev_df.rename(columns={'vin_(1-10)':'VIN'},inplace=True)


ev_df.head(5)
```

```python
#checking the data types of columns
ev_df.dtypes

#Correcting the data types of required columns
ev_df['postal_code'] = ev_df['postal_code'].astype('Int64')
ev_df['legislative_district'] = ev_df['legislative_district'].astype('Int64')

ev_df.dtypes
```

```python
#Creating new calculated columns

#Creating current year and calculating vehicle age from it
ev_df['current_year']=pd.Timestamp.now().year

ev_df['model_year'] = pd.to_numeric(ev_df['model_year'])
ev_df['vehicle_age']=ev_df['current_year']-ev_df['model_year']

ev_df['vehicle_age']
```

```python
ev_df['high_range_ev'] = ev_df['electric_range'].apply(lambda x: 'yes' if x > 200 else 'no')

ev_df.head(5)
```

```python
#checking all the states values in DataFrame
unique_states=ev_df['state'].unique()
print(unique_states)

#checking number of rows for each state
count_states=ev_df['state'].value_counts()
print(count_states)

#filtering only Washington state
ev_df=ev_df[ev_df['state']=='WA']

ev_df.shape
```

```python
#dropping irrelevant columns
ev_df=ev_df.drop(columns=['2020_census_tract','current_year'])

ev_df
```

```python
#checking for null values
null_counts = ev_df.isnull().sum()

print(null_counts)


size= ev_df.shape

print(f"The size of DataFrame is :{size}")

#there aren't much null values in the entire dataset range
```

```python
#check for duplicate rows in the DataFrame
# Count the number of duplicate rows
num_duplicates = ev_df.duplicated().sum()

print(num_duplicates)
```
